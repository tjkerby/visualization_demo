# 0. Instructor opener: what do the summaries hide?
Four datasets have nearly identical means, variances, correlations, and fitted regression lines. Their scatterplots reveal very different structures.

**Teaching sequence:** show the summary table first, ask whether similar summaries imply similar patterns, then reveal the four plots. The table-building code is preparation; there is no need to teach its syntax in this visualization lesson.

This opener is optional. Begin with `1-matplotlib.ipynb` if the idea is already familiar.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Anscombe's quartet: four deliberately constructed datasets, 11 points each.
x_shared = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y_values = [
    [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68],
    [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74],
    [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73],
    [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89],
]
quartet = pd.concat([
    pd.DataFrame({"dataset": i + 1,
                  "x": x_shared if i < 3 else [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8],
                  "y": y})
    for i, y in enumerate(y_values)
], ignore_index=True)
quartet.head()
```

```python
rows = []
for name, group in quartet.groupby("dataset"):
    slope, intercept = np.polyfit(group["x"], group["y"], 1)
    rows.append({"dataset": name, "mean_x": group["x"].mean(),
                 "mean_y": group["y"].mean(), "variance_x": group["x"].var(),
                 "variance_y": group["y"].var(),
                 "correlation": group["x"].corr(group["y"]),
                 "slope": slope, "intercept": intercept})
summary = pd.DataFrame(rows).set_index("dataset")
summary.round(3)
```

## Before the reveal
The summaries describe some aspects of the data while discarding others. A useful question for the room: could a curved pattern or one influential observation be hiding behind these numbers?


```python
fig, axes = plt.subplots(2, 2, figsize=(9, 7), sharex=True, sharey=True, layout="constrained")
for ax, (name, group) in zip(axes.flat, quartet.groupby("dataset")):
    ax.scatter(group["x"], group["y"], color="#0072B2", s=45)
    slope, intercept = np.polyfit(group["x"], group["y"], 1)
    x_line = np.array([3, 20])
    ax.plot(x_line, intercept + slope * x_line, color="#D55E00", linewidth=1.5)
    ax.set(title=f"Dataset {name}", xlabel="x", ylabel="y", xlim=(2, 21), ylim=(2, 14))
fig.suptitle("Similar summaries, different structures")
plt.show()
```

## What the plots reveal
- Dataset 1: a roughly linear relationship with scatter.
- Dataset 2: a curved relationship.
- Dataset 3: an unusual y value beside a roughly linear main pattern.
- Dataset 4: one high-leverage point strongly determines the fitted line.

Shared axis ranges make the panels easier to compare. Numerical summaries and visual inspection complement each other.
