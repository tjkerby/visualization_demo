# 2. Seaborn: let columns describe the plot
**Instructor demonstration.** Use the same data and scatterplot so the differences from Matplotlib are visible. Add one mapping at a time, then show how Seaborn summarizes a distribution.

Seaborn builds on Matplotlib. Its axes-level functions return a Matplotlib `Axes`, so methods such as `set_xscale` and `set` still work.


```python
import pandas as pd
import matplotlib.pyplot as plt

gapminder = pd.read_csv("data/gapminder.csv")
snapshot = gapminder.loc[gapminder["year"] == 2007].copy()

import seaborn as sns
```

## 1. The simplest scatterplot
Unlike Matplotlib's `ax.scatter(x_values, y_values)`, Seaborn accepts `data=` and the names of the columns. It supplies axis labels from those names.


```python
ax = sns.scatterplot(data=snapshot, x="gdpPercap", y="lifeExp")
plt.show()
```

## 2. Add one grouping variable
**New:** `hue="continent"` maps continent to color and creates a legend. Seaborn handles grouping the rows for us.

Continent is categorical. Distinct hues communicate group identity; a light-to-dark scale would suggest an order that is not present.


```python
ax = sns.scatterplot(data=snapshot, x="gdpPercap", y="lifeExp", hue="continent")
plt.show()
```

## 3. Use the Matplotlib methods we already learned
**New:** a log scale and readable labels. Seaborn creates the marks and legend; Matplotlib refines the plotting area.


```python
ax = sns.scatterplot(data=snapshot, x="gdpPercap", y="lifeExp", hue="continent")
ax.set_xscale("log")
ax.set(xlabel="GDP per capita (inflation-adjusted US$); log scale",
       ylabel="Life expectancy at birth (years)", title="Countries by continent, 2007")
plt.show()
```

## 4. Start a distribution comparison with a boxplot
**New plot:** `sns.boxplot(...)` computes and draws a summary of each group.

The line is the median, and the box contains the middle 50%. Whiskers reach the most extreme values within 1.5 IQR of the box; more distant observations are shown as separate points. This is a statistical summary, not just a different drawing style.


```python
ax = sns.boxplot(data=snapshot, x="continent", y="lifeExp")
plt.show()
```

## 5. Reveal the observations behind the summaries
**New:** `sns.stripplot(...)` adds individual countries to the same `ax`. `jitter=True` spreads overlapping categorical positions horizontally without changing the data values.

The neutral box color makes the points easier to see. `showfliers=False` hides the boxplot's separate outlier markers because every observation is already shown by the strip plot.


```python
ax = sns.boxplot(data=snapshot, x="continent", y="lifeExp", color="lightgray", showfliers=False)
sns.stripplot(data=snapshot, x="continent", y="lifeExp", color="black", jitter=True, ax=ax)
ax.set(xlabel="Continent", ylabel="Life expectancy at birth (years)", title="Countries in 2007")
plt.show()
```

## 6. Check how many observations support each summary
Only two countries represent Oceania in this extract. A polished boxplot does not guarantee a large or representative sample. The individual points make that limitation visible.


```python
snapshot.groupby("continent")["lifeExp"].agg(["count", "mean", "median"]).round(1)
```

## Advanced showcase: compare two years with small multiples
Now combine familiar column mappings with one new idea: **one panel per year**. `col="year"` creates the panels, while `hue` and `style` map continent to color and marker shape.

`relplot` manages the entire figure and returns a grid, unlike the axes-level functions above. Its `height` and `aspect` control panel size. Shared, fixed axes make the two years easier to compare. Each point is one country in the panel's year.


```python
endpoints = gapminder.loc[gapminder["year"].isin([1952, 2007])]

g = sns.relplot(data=endpoints, x="gdpPercap", y="lifeExp",
                col="year", hue="continent", style="continent",
                palette="colorblind", s=55, alpha=0.75,
                height=4, aspect=1.1)
g.set(xscale="log", xlim=(200, 120000), ylim=(20, 90))
g.set_axis_labels("GDP per capita (US$); log scale", "Life expectancy at birth (years)")
g.set_titles("Countries in {col_name}")
g.figure.suptitle("Compare the same countries at two points in time", y=1.03)
from pathlib import Path
Path("outputs").mkdir(exist_ok=True)
g.figure.savefig("outputs/seaborn_advanced.png", dpi=200, bbox_inches="tight")
plt.show()
```

**What this shows off:** Seaborn handles grouping, redundant visual mappings, the legend, and a repeated panel layout from DataFrame columns. Color and shape meanings are consistent between these two panels. We still use Matplotlib's `savefig` on the underlying figure.

These two snapshots reveal differences between the endpoints, while hiding changes in between. The Plotly animation will let us inspect those intermediate years.

Reference: [Seaborn relational plots and facets](https://seaborn.pydata.org/generated/seaborn.relplot.html).


## What Seaborn contributes
**Strength:** concise statistical graphics from a DataFrame. It handles group mappings, legends, and summaries, while preserving access to Matplotlib customization.

**Tradeoff:** convenient functions make choices for us. A boxplot summarizes a distribution; `barplot` computes a mean by default, while `countplot` counts rows. Understand the statistic rather than assuming a function draws raw values.
