# 1. Matplotlib: build a figure one step at a time
**Instructor demonstration.** Start with the simplest working scatterplot. Add labels, change the scale, then improve the marks. Each version repeats the small plot so the added arguments are easy to identify.

**Question:** How are GDP per capita and life expectancy related across countries in 2007?

One row of `gapminder` is one country-year. One row of `snapshot` is one country in 2007. GDP per capita is in inflation-adjusted US dollars; life expectancy at birth is in years. These historical teaching data cover 142 countries, not every country. See [data notes](data/README.md).


```python
import pandas as pd
import matplotlib.pyplot as plt

gapminder = pd.read_csv("data/gapminder.csv")
snapshot = gapminder.loc[gapminder["year"] == 2007].copy()
```

## 1. The smallest useful starting point
`fig` is the whole figure. `ax` is one plotting area, including its axes. `ax.scatter(x, y)` places the observations.

Matplotlib accepts arrays or columns directly. It does not automatically know the variable names or units we want to show.


```python
fig, ax = plt.subplots()
ax.scatter(snapshot["gdpPercap"], snapshot["lifeExp"])
plt.show()
```

## 2. Add labels and a title
**New:** `ax.set(...)` supplies context. The points are unchanged.


```python
fig, ax = plt.subplots()
ax.scatter(snapshot["gdpPercap"], snapshot["lifeExp"])
ax.set(xlabel="GDP per capita (inflation-adjusted US$)",
       ylabel="Life expectancy at birth (years)", title="Countries in 2007")
plt.show()
```

## 3. Make the crowded part easier to see
**New:** `ax.set_xscale("log")`. The values have not changed; the distances representing them have.

On a log axis, equal distances represent equal ratios. For example, 1,000 to 10,000 occupies the same distance as 10,000 to 100,000. All GDP values here are positive. The y-axis remains linear.


```python
fig, ax = plt.subplots()
ax.scatter(snapshot["gdpPercap"], snapshot["lifeExp"])
ax.set(xlabel="GDP per capita (inflation-adjusted US$); log scale",
       ylabel="Life expectancy at birth (years)", title="Countries in 2007")
ax.set_xscale("log")
plt.show()
```

## 4. Adjust the marks when they overlap
**New:** `s` controls marker area and `alpha` controls opacity. Smaller, partly transparent points can make overlap easier to see. These are design choices, not requirements for every scatterplot.


```python
fig, ax = plt.subplots()
ax.scatter(snapshot["gdpPercap"], snapshot["lifeExp"], s=30, alpha=0.65)
ax.set(xlabel="GDP per capita (inflation-adjusted US$); log scale",
       ylabel="Life expectancy at birth (years)", title="Countries in 2007")
ax.set_xscale("log")
plt.show()
```

## 5. Add a second plotting area
**New:** `plt.subplots(1, 2)` creates two panels. `sharey=True` keeps their y scales aligned. The familiar plotting commands now go to `axes[0]` and `axes[1]`.

The figure size gives both panels room. `tight_layout()` helps keep labels from overlapping. This is where Matplotlib's explicit control starts to pay off.


```python
fig, axes = plt.subplots(1, 2, sharey=True, figsize=(11, 4))
axes[0].scatter(snapshot["gdpPercap"], snapshot["lifeExp"])
axes[1].scatter(snapshot["gdpPercap"], snapshot["lifeExp"])
axes[1].set_xscale("log")
axes[0].set(xlabel="GDP per capita (US$)", ylabel="Life expectancy (years)", title="Linear x scale")
axes[1].set(xlabel="GDP per capita (US$)", title="Logarithmic x scale")
fig.tight_layout()
plt.show()
```

## 6. Save the figure
`fig.savefig(...)` saves the Matplotlib figure. PNG is useful for a report or webpage; SVG preserves vector graphics. This cell saves the two-panel comparison above.


```python
from pathlib import Path

Path("outputs").mkdir(exist_ok=True)
fig.savefig("outputs/linear_vs_log.png", dpi=200, bbox_inches="tight")
```

## Advanced showcase: shape, color, and a targeted annotation
The basic plot is complete. This optional final example shows how Matplotlib can combine the choices we learned with **different marker symbols** and a precisely placed annotation.

Color and shape both identify continent. The redundant shape cue helps when colors are difficult to distinguish. We draw each group explicitly and use `marker=` to choose its symbol. All markers have the same area; size does not encode another variable here.

The arrow identifies Japan without labeling every country. This is a demonstration of control, rather than a list of arguments to memorize.


```python
markers = ["o", "s", "^", "D", "P"]
colors = ["#0072B2", "#D55E00", "#009E73", "#CC79A7", "#9A6700"]

fig, ax = plt.subplots(figsize=(9, 5))
for (continent, group), marker, color in zip(snapshot.groupby("continent"), markers, colors):
    ax.scatter(group["gdpPercap"], group["lifeExp"], label=continent,
               marker=marker, color=color, s=65, alpha=0.75)

ax.set_xscale("log")
ax.set(xlabel="GDP per capita (inflation-adjusted US$); log scale",
       ylabel="Life expectancy at birth (years)", ylim=(35, 90),
       title="Country life expectancy and GDP per capita, 2007")
ax.legend(title="Continent", frameon=False, loc="lower right")
ax.grid(axis="y", alpha=0.2)

japan = snapshot.loc[snapshot["country"] == "Japan"].iloc[0]
ax.annotate("Japan", xy=(japan["gdpPercap"], japan["lifeExp"]),
            xytext=(-45, 20), textcoords="offset points",
            arrowprops={"arrowstyle": "->", "color": "black"})
fig.tight_layout()
fig.savefig("outputs/matplotlib_advanced.png", dpi=200, bbox_inches="tight")
plt.show()
```

**What this shows off:** Matplotlib gives us control of the symbol, appearance, legend, and annotation for each group. In Seaborn, we will see how named-column mappings can automate much of the grouping.

Reference: [Matplotlib marker symbols](https://matplotlib.org/stable/api/markers_api.html).


## What Matplotlib contributes
**Strength:** direct control over marks, annotations, axes, and layouts. It is useful for a carefully composed static figure.

**Tradeoff:** grouping observations and building statistical summaries often take more code. It can support interactive backends, but notebook hover labels and browser interactions are not the focus of these examples.

**Interpretation:** life expectancy tends to be higher among countries with higher GDP per capita. This country-level observational comparison does not establish causation.
